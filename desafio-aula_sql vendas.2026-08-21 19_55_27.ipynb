```python
%sql
-- 1. RECEITA BRUTA TOTAL
SELECT SUM(Receita) AS Receita_Bruta_Total 
FROM vendas;
```

```python
%sql
-- 2. RENTABILIDADE BRUTA (LUCRO)
SELECT 
    SUM(Receita - Custo) AS Lucro_Bruto_Total 
FROM vendas;
```

```python
%sql
-- 3. MARGEM DE LUCRO PERCENTUAL (%)
SELECT 
    (SUM(Receita - Custo) / SUM(Receita)) * 100 AS Margem_Lucro_Percentual 
FROM vendas;
```

```python
%sql
-- 4. VOLUME TOTAL VENDIDO
SELECT 
    SUM(Quantidade) AS Volume_Total_Vendido 
FROM vendas;
```

```python
%sql
-- 5. RECEITA POR CATEGORIA COMERCIAL
SELECT 
    Categoria, 
    SUM(Receita) AS Receita_Total 
FROM vendas 
GROUP BY Categoria 
ORDER BY Receita_Total DESC;
```

```python
%sql
-- 6. RECEITA DISTRIBUÍDA POR ESTADO (UF)
SELECT 
    Estado, 
    SUM(Receita) AS Receita_Total 
FROM vendas 
GROUP BY Estado 
ORDER BY Receita_Total DESC;
```

```python
%sql
-- 7. Evolução Temporal do Faturamento
SELECT 
    DATE_TRUNC('month', Data) AS Mes, 
    SUM(Receita) AS Receita_Mensal 
FROM vendas 
GROUP BY DATE_TRUNC('month', Data) 
ORDER BY Mes ASC;
```

```python
%sql
-- 8. PERFORMANCE: TOP VENDEDORES
SELECT 
    Vendedor, 
    SUM(Receita) AS Receita_Total 
FROM vendas 
GROUP BY Vendedor 
ORDER BY Receita_Total DESC 
LIMIT 10;
```

```python
%sql
-- 9. PARTICIPAÇÃO POR CANAL DE VENDA
SELECT 
    `Canal de Venda`, 
    SUM(Receita) AS Receita_Total,
    (SUM(Receita) / SUM(SUM(Receita)) OVER()) * 100 AS Participacao_Percentual
FROM vendas 
GROUP BY `Canal de Venda` 
ORDER BY Receita_Total DESC;
```
