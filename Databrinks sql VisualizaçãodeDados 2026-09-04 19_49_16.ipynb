1. Performance comercial 

```python
%sql
-- Top 5 vendedores por Receita
SELECT 
    Vendedor,
    SUM(Receita) AS receita_total
FROM workspace.default.vendas
GROUP BY Vendedor
ORDER BY receita_total DESC
LIMIT 5;
```

```python
%sql
-- Top 5 vendedores por Lucro
SELECT 
    Vendedor,
    SUM(Lucro) AS lucro_total
FROM workspace.default.vendas
GROUP BY Vendedor
ORDER BY lucro_total DESC
LIMIT 5;
```

Ao analisar os dois rankings, é possível notar que os cinco vendedores que aparecem nas primeiras posições são os mesmos, mas eles não estão exatamente na mesma ordem. A mudança principal aconteceu entre a 3ª e a 4ª colocação: a vendedora Isabela Santos gerou mais receita e ficou em 3º lugar em faturamento, mas caiu para a 4ª posição no ranking de lucro. Já a vendedora Eduarda Alves, mesmo faturando menos no total, superou a Isabela e assumiu o 3º lugar em lucro. Na prática do negócio, essa diferença mostra que vender mais em volume financeiro nem sempre significa trazer o melhor resultado financeiro para a empresa. Isso costuma acontecer quando um vendedor oferece descontos excessivos para fechar a venda ou foca em itens que possuem margem de ganho menor, enquanto outro vendedor prioriza produtos mais rentáveis ou preserva o preço original.

PARTE 2 - Caso profissional 

Etapa A - Detectar o problema

```python
%sql
-- Etapa A: Evolução de Receita, Lucro e Margem por Mês
SELECT 
    DATE_TRUNC('month', Data) AS mes,
    ROUND(SUM(Receita), 2) AS receita,
    ROUND(SUM(Lucro), 2) AS lucro,
    ROUND((SUM(Lucro) / SUM(Receita)) * 100, 2) AS margem_pct
FROM workspace.default.vendas
GROUP BY DATE_TRUNC('month', Data)
ORDER BY mes;
```

1. Receita e lucro evoluem igual? 
Não. Em vários meses a receita se mantém alta ou sobe, mas o lucro não cresce no mesmo ritmo. 

2. Receita sobe e margem cai?
Sim, ao analisar a coluna margem_pct, percebe-se meses em que o faturamento aumentou, mas a margem percentual diminuiu.  

3. Melhor e pior período:
O mês de maior receita foi agosto de 2025 (pico de faturamento) e o pior foi março de 2025 (menor receita histórica).

4. Pontual ou tendência? 
É uma oscilação recorrente com tendência de achatamento do lucro nos meses de promoções ou descontos mais agressivos. 

Etapa B - Encontrar possíveis causas

```python
%sql
-- Cruzamento 1: Categoria x Canal
SELECT 
    Categoria,
    `Canal de Venda`,
    ROUND(SUM(Receita), 2) AS receita,
    ROUND(SUM(Lucro), 2) AS lucro,
    ROUND((SUM(Lucro) / SUM(Receita)) * 100, 2) AS margem_pct
FROM workspace.default.vendas
GROUP BY Categoria, `Canal de Venda`
ORDER BY Categoria, receita DESC;
```

```python
%sql
-- Cruzamento 2: Categoria x Estado
SELECT 
    Categoria,
    Estado,
    ROUND(SUM(Receita), 2) AS receita,
    ROUND(SUM(Lucro), 2) AS lucro,
    ROUND((SUM(Lucro) / SUM(Receita)) * 100, 2) AS margem_pct
FROM workspace.default.vendas
GROUP BY Categoria, Estado
ORDER BY Categoria, receita DESC;
```

```python
%sql
-- Análise de Rentabilidade por Categoria
SELECT 
    Categoria,
    ROUND(SUM(Receita), 2) AS receita,
    ROUND(SUM(Lucro), 2) AS lucro,
    ROUND((SUM(Lucro) / SUM(Receita)) * 100, 2) AS margem_pct
FROM workspace.default.vendas
GROUP BY Categoria
ORDER BY margem_pct ASC;
```

Etapa C - Encontrar uma anomalia ou oportunidade


A anomalia encontrada é que a categoria de Informática, apesar de liderar o faturamento com mais de R$ 17 milhões e concentrar as vendas em São Paulo e nos canais digitais, opera com uma margem de lucro proporcionalmente mais baixa devido a custos altos ou descontos excessivos para fechar negócios. Em contrapartida, a grande oportunidade está em produtos menores, como Acessórios e Escritório, além de praças e canais com menor volume, que entregam margens percentuais muito superiores; assim, a empresa pode aplicar estratégias de venda cruzada oferecendo acessórios junto aos itens de informática para alavancar a rentabilidade final e o retorno financeiro sem depender apenas de vender produtos caros com margens apertadas.
